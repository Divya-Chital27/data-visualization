**Importing Libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Importing Datasets for analysis**

```python
tips=sns.load_dataset('tips')
iris=sns.load_dataset('iris')

#a small synthetic dataset of two batsmen's runs across seasons, used
#throughout the line plot section
seasons=[2015,2016,2017,2018,2019,2020]
rohit_runs=[1250,900,1400,1100,1500,1050]
virat_runs=[1000,1200,1350,1450,1300,1600]
```

**What is Matplotlib**
<span>

Matplotlib is the foundational Python data visualization library.
It provides fine-grained, low-level control over every element of a
figure-lines, markers, axes, ticks, legends-making it possible to
build almost any static chart. Most other visualization libraries,
including Seaborn, are built on top of it </span>

**Types of Matplotlib Interfaces**


*   Pyplot Interface (plt)-<span>A state-based, MATLAB-style interface where each function call (plt.plot, plt.title, etc.) acts on the current figure/axes. Quick and convenient for simple, single-plot use cases.</span>


 - plt.plot()
 - plt.scatter()
 - plt.bar()
 - plt.barh()
 - plt.hist()
 - plt.pie()
 - plt.xlabel() / plt.ylabel() / plt.title()
 - plt.legend()
 - plt.show()



*   Object-Oriented Interface (OO)-<span>Explicitly creates Figure and Axes objects and calls methods on them. Gives full control over complex, multi-panel figures and is the recommended approach for anything beyond a single simple plot.</span>


- plt.subplots()
 - fig.add_subplot()
 - ax.scatter() / ax.plot() / ax.hist()
 - ax.set_title() / ax.set_xlabel()
 - fig.colorbar()






**Line** **Plots**

Pyplot Interface-plt.plot() draws a line connecting a series of points, ideal for ordered/continuous data.

*   Simple Line Plot
*   Multiple Lines
*   Styling (color, linestyle, linewidth, marker, legend)
*   Limiting Axes
*   Grid




```python
#Simple line plot
# Purpose: Shows trends or changes over an ordered variable such as time.
# Best Used For:
# - Time series analysis
# - Trend analysis
# - Comparing multiple trends
# Insight:
# The price shows a general upward trend from 2015 to 2017, followed
# by a decline through 2020, suggesting the value peaked in the middle
# of the period before falling.
price=[48000,54000,57000,49000,47000,45000]
year=[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
```

```python
#Multiple lines on the same plot
# Purpose:
# Plotting more than one line on the same axes allows a direct,
# point-by-point comparison between two series.
# Insight:
# Virat's run tally overtakes Rohit's in the later seasons, even
# though Rohit had a stronger start in 2017.
plt.plot(seasons,rohit_runs)
plt.plot(seasons,virat_runs)
```

```python
#title, xlabel, ylabel
# Purpose:
# Adds descriptive text to the figure so the chart can be
# understood without additional context.
plt.plot(seasons,rohit_runs)
plt.plot(seasons,virat_runs)
plt.title('Rohit Sharma Vs Virat Kohli')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
#color and linestyle
# color-Sets the line color.
# linestyle-Controls whether the line is solid, dashed, or dotted.
plt.plot(seasons,rohit_runs,color='green',linestyle='dashed')
plt.plot(seasons,virat_runs,color='black',linestyle='dotted')
plt.title('Rohit Sharma Vs Virat Kohli')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
#linewidth
# Purpose:
# Controls the thickness of the plotted line, useful for
# emphasizing one series over another.
plt.plot(seasons,rohit_runs,color='green',linestyle='solid',linewidth=3)
plt.plot(seasons,virat_runs,color='black',linestyle='dotted',linewidth=2)
plt.title('Rohit Sharma Vs Virat Kohli')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
#marker
# Purpose:
# Adds a symbol at each data point, making individual observations
# on the line easier to spot.
plt.plot(seasons,rohit_runs,color='green',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(seasons,virat_runs,color='black',linestyle='dotted',linewidth=2,marker='D')
plt.title('Rohit Sharma Vs Virat Kohli')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
#legend
# Purpose:
# Displays a key that maps each line's label to its color/style,
# essential when multiple series are plotted together.
plt.plot(seasons,rohit_runs,color='green',linestyle='solid',linewidth=3,marker='D',label='Rohit')
plt.plot(seasons,virat_runs,color='black',linestyle='dotted',linewidth=2,marker='D',label='Virat')
plt.title('Rohit Sharma Vs Virat Kohli')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend(loc='upper left')
```

```python
#limiting axes
# Purpose:
# ylim()/xlim() restrict the visible range of an axis, which is
# useful when a single outlier would otherwise compress the rest
# of the data.
# Insight:
# Without a limit, the 2021 outlier value flattens every other
# year into a nearly straight line near zero. Setting ylim keeps
# the earlier years readable at the cost of clipping the outlier.
price=[48000,54000,57000,49000,47000,45000,4500000]
year=[2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
plt.ylim(0,100000)
```

```python
#grid
# Purpose:
# Adds horizontal and vertical gridlines, making it easier to
# read exact values off the axes.
plt.plot(seasons,rohit_runs,color='green',linestyle='solid',linewidth=3,marker='D',label='Rohit')
plt.plot(seasons,virat_runs,color='black',linestyle='dotted',linewidth=2,marker='D',label='Virat')
plt.title('Rohit Sharma Vs Virat Kohli')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend(loc='upper left')
plt.grid()
plt.show()
```

**Scatter** **Plots**


*   Basic Scatter Plot
*   Styling (color, marker)
*   Size Parameter
*   Scatter via plt.plot()





```python
#Basic scatter plot
# Purpose: Shows the relationship between two numerical variables
# as individual points.
# Best Used For:
# - Finding correlations
# - Detecting clusters
# - Identifying outliers
# Insight:
# The points roughly follow an upward linear trend with noticeable
# scatter around it, consistent with y being a noisy linear
# function of x.
x=np.linspace(-10,10,50)
y=10*x+3+np.random.randint(0,300,50)
plt.scatter(x,y)
```

```python
#color and marker
# Insight:
# Customers with higher total bills generally leave higher tips,
# and using a '+' marker keeps overlapping points easy to
# distinguish.
plt.scatter(tips['total_bill'],tips['tip'],color='red',marker='+')
plt.title('total bill vs tip')
plt.xlabel('total bill')
plt.ylabel('tip')
```

```python
#size (s parameter)
# Purpose:
# The s parameter scales the marker size by a variable, turning
# the scatter plot into a bubble chart.
# Insight:
# Larger points (bigger dining parties) tend to cluster at higher
# total bill values, which is expected since bigger groups order more.
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```

```python
#scatter plot using plt.plot
# Purpose:
# plt.plot() can also produce a scatter-style plot by using a
# marker style ('o') with no connecting line, and is slightly
# faster than plt.scatter() for large datasets.
plt.plot(tips['total_bill'],tips['tip'],'o')
```

**Bar** **Plots**


*   Simple Bar Chart
*   Horizontal Bar Chart
*   Side-by-Side (Grouped) Bar Chart
*   Stacked Bar Chart





```python
#Simple bar chart
# Purpose:
# Compares values across categories using the height of bars.
# Best Used For:
# - Comparing quantities across categories
# - Ranking categories
child=[10,20,40,10,30]
color=['red','blue','green','yellow','pink']
plt.bar(color,child,color='black')
```

```python
#Horizontal bar chart
# Purpose:
# Same comparison as a vertical bar chart but drawn sideways,
# useful when category labels are long.
plt.barh(color,child,color='black')
```

```python
#Side-by-side (grouped) bar chart
# Purpose:
# Places bars for different groups next to each other at each
# category position instead of overlapping them, making direct
# comparison across groups easy.
# Insight:
# Both batsmen show year-to-year fluctuation, but placing the
# bars side by side makes it easy to see which player scored
# more in each individual season.
runs_df=pd.DataFrame({'batsman':['Rohit','Virat'],
                       '2015':[1250,1000],
                       '2016':[900,1200],
                       '2017':[1400,1350]})
plt.bar(np.arange(runs_df.shape[0])-0.2,runs_df['2015'],width=0.2,label='2015')
plt.bar(np.arange(runs_df.shape[0]),runs_df['2016'],width=0.2,label='2016')
plt.bar(np.arange(runs_df.shape[0])+0.2,runs_df['2017'],width=0.2,label='2017')
plt.xticks(np.arange(runs_df.shape[0]),runs_df['batsman'])
plt.legend()
```

```python
#long category labels
# Purpose:
# rotation='vertical' rotates x-axis tick labels, preventing long
# category names from overlapping each other.
child=[10,20,40,10,30]
color=['red red red red red red','blue blue blue blue blue blue','green green green green green green green','yellow','pink']
plt.bar(color,child,color='black')
plt.xticks(rotation='vertical')
```

```python
#Stacked bar chart
# Purpose:
# Stacks bars for sub-categories (here, seasons) on top of each
# other using the bottom parameter, showing both the total and
# each season's individual contribution.
# Insight:
# The combined bar height represents each batsman's total runs
# across the three seasons, while each colored segment shows how
# much of that total came from a specific season.
plt.bar(runs_df['batsman'],runs_df['2015'],label='2015')
plt.bar(runs_df['batsman'],runs_df['2016'],bottom=runs_df['2015'],label='2016')
plt.bar(runs_df['batsman'],runs_df['2017'],bottom=(runs_df['2016']+runs_df['2015']),label='2017')
plt.legend()
```

**Histogram & Pie Charts**


*   Histogram
*   Logarithmic Histogram
*   Pie Chart
*   Pie Chart with Percentages
*   Exploded Pie Chart
*   Pie Chart with Shadow





```python
#Histogram
# Purpose:
# Displays the distribution of a single numerical variable by
# grouping data into intervals (bins) and showing the frequency
# of observations in each interval.
# Best Used For:
# - Understanding the distribution of data
# - Detecting skewness
# - Identifying the most common value ranges
# Insight:
# Most total bill values fall in the lower to middle bins, and the
# number of observations decreases as the bill amount increases,
# indicating a right-skewed distribution.
plt.hist(tips['total_bill'],bins=[0,10,20,30,40,50,60])
```

```python
#Logarithmic histogram
# Purpose:
# log=True plots the y-axis (frequency) on a log scale, which is
# useful when a few bins have very high counts that would
# otherwise dwarf the rest of the bars.
plt.hist(tips['size'],bins=[1,2,3,4,5,6,7],log=True)
```

```python
#Pie chart
# Purpose:
# Shows the proportion of each category relative to the whole
# using slices of a circle.
# Best Used For:
# - Showing simple part-to-whole relationships
# - Comparing the share of a small number of categories
data=[23,45,100,20,49]
sub=['eng','sci','maths','sst','hindi']
plt.pie(data,labels=sub)
plt.show()
```

```python
#autopct
# Purpose:
# Displays the percentage share of each slice directly on the
# pie chart.
# Insight:
# Maths makes up the single largest share of the total marks
# among all five subjects, while English and SST contribute the
# smallest shares.
plt.pie(data,labels=sub,autopct='%0.1f%%')
plt.show()
```

```python
#explode
# Purpose:
# Pulls one or more slices out from the center of the pie to
# visually highlight them.
plt.pie(data,labels=sub,autopct='%0.1f%%',explode=[0.1,0,0,0,0])
plt.show()
```

```python
#shadow
# Purpose:
# Adds a drop shadow beneath the pie chart for a more
#3D-styled appearance.
plt.pie(data,labels=sub,autopct='%0.1f%%',shadow=True)
plt.show()
```

**Styling & Saving Figures**


*   Built-in Styles
*   Saving a Figure





```python
#Built-in styles
# Purpose:
# Matplotlib ships with several pre-built visual themes that
# change fonts, colors, and gridlines with a single call.
# plt.style.available lists every available style name.
print(plt.style.available[:10])
plt.style.use('grayscale')
plt.pie(data,labels=sub,autopct='%0.1f%%',shadow=True)
plt.show()
```

```python
#Saving a figure
# Purpose:
# savefig() writes the current figure to disk as an image file
# (png, jpg, pdf, svg, etc.), useful for exporting charts for
# reports or presentations.
plt.pie(data,labels=sub,autopct='%0.1f%%',shadow=True)
plt.savefig('pie_chart.png')
```

**Annotations & Reference Lines**


*   Colorbar with Color Mapping
*   Figure Size
*   Text Annotations
*   Horizontal & Vertical Reference Lines





```python
#colorbar with c and cmap
# Purpose:
# The c parameter colors each point according to a numeric
# variable, and cmap chooses the color gradient. colorbar()
# adds a legend explaining the color scale.
# Insight:
# The three iris species form fairly distinct color bands across
# sepal length and petal width, especially separating setosa from
# the other two species.
species_num=iris['species'].map({'setosa':0,'versicolor':1,'virginica':2})
plt.scatter(iris['sepal_length'],iris['petal_width'],c=species_num,cmap='winter',alpha=0.7)
plt.xlabel('sepal length')
plt.ylabel('petal width')
plt.colorbar()
```

```python
#figure size
# Purpose:
# figsize controls the width and height of the figure in inches,
# useful for making crowded plots more readable.
plt.figure(figsize=(10,6))
plt.scatter(iris['sepal_length'],iris['petal_width'],c=species_num,cmap='winter',alpha=0.7)
plt.xlabel('sepal length')
plt.ylabel('petal width')
plt.colorbar()
```

```python
#text annotations
# Purpose:
# plt.text() places a text label at a specific (x, y) coordinate,
# useful for labeling individual points directly on the chart.
sample_df=tips.sample(15,random_state=5).reset_index(drop=True)
plt.figure(figsize=(10,7))
plt.scatter(sample_df['total_bill'],sample_df['tip'],s=sample_df['size']*20)
for i in range(sample_df.shape[0]):
    plt.text(sample_df['total_bill'][i],sample_df['tip'][i],sample_df['day'][i])
```

```python
#horizontal and vertical reference lines
# Purpose:
# axhline()/axvline() draw a full-width horizontal or full-height
# vertical line at a given value, useful for marking thresholds
# or averages on top of a scatter plot.
# Insight:
# The reference lines divide customers into four groups based on
# whether their bill and tip are above or below the chosen
# thresholds, making it easy to spot high-tip, low-bill outliers.
plt.figure(figsize=(10,7))
plt.scatter(tips['total_bill'],tips['tip'])
plt.axhline(4,color='red')
plt.axvline(20,color='red')
```

**Subplots (Object-Oriented API)**


*   Single Axes via subplots()
*   Multiple Rows (sharex)
*   Grid of Subplots (2x2)
*   add_subplot()





```python
#Single Axes via plt.subplots()
# Purpose:
# The object-oriented alternative to plt.scatter()/plt.plot();
# fig and ax are created explicitly and methods are called on ax,
# giving more control for complex figures.
fig,ax=plt.subplots(figsize=(8,5))
ax.scatter(tips['total_bill'],tips['tip'],color='red')
ax.set_title('total bill vs tip')
ax.set_xlabel('total bill')
ax.set_ylabel('tip')
```

```python
#Multiple rows with shared x-axis
# Purpose:
# sharex=True links the x-axis across all subplots, so zooming
# or panning one updates the others, and keeps them visually aligned.
# Insight:
# Stacking the two plots with a shared x-axis makes it easy to
# compare how tip and party size both vary with total bill at the
# same x position.
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(8,6))
ax[0].scatter(tips['total_bill'],tips['tip'],color='red')
ax[1].scatter(tips['total_bill'],tips['size'])
ax[0].set_title('total bill vs tip')
ax[1].set_title('total bill vs size')
ax[0].set_ylabel('tip')
ax[1].set_xlabel('total bill')
ax[1].set_ylabel('size')
```

```python
#Grid of subplots (2x2)
# Purpose:
# nrows and ncols together create a grid of independent axes,
# each of which can hold a different chart type, useful for
# building a small dashboard in one figure.
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,8))
ax[0,0].scatter(tips['total_bill'],tips['tip'],color='red')
ax[0,1].scatter(tips['total_bill'],tips['size'])
ax[1,0].hist(tips['total_bill'])
ax[1,1].hist(tips['tip'])
```

```python
#add_subplot()
# Purpose:
# An alternative way to build multi-panel figures by adding one
# axes at a time to a figure using a 3-digit position code
# (rows, columns, index).
fig=plt.figure()
ax1=fig.add_subplot(211)
ax1.scatter(tips['total_bill'],tips['tip'])
ax2=fig.add_subplot(212)
ax2.scatter(tips['total_bill'],tips['size'])
```

**3D Plots**


*   3D Scatter Plot
*   3D Line Plot
*   3D Surface Plot





```python
#3D scatter plot
# Purpose:
# Extends a regular scatter plot with a third numerical axis,
# allowing three variables to be visualized simultaneously.
# projection='3d' turns a normal axes into a 3D axes.
sample_df=tips.sample(30,random_state=5)
fig=plt.figure()
ax=plt.subplot(projection='3d')
ax.scatter3D(sample_df['total_bill'],sample_df['tip'],sample_df['size'])
ax.set_title('total bill vs tip vs size')
```

```python
#3D line plot
# Purpose:
# Draws a connected line through points in three-dimensional
# space rather than a flat 2D plane.
x=[0,1,5]
y=[0,10,13]
z=[0,13,20]
fig=plt.figure()
ax=plt.subplot(projection='3d')
ax.plot3D(x,y,z)
ax.set_title('simple 3d line plot')
```

```python
#3D surface plot
# Purpose:
# Displays a continuous surface over a two-dimensional grid,
# useful for visualizing mathematical functions or terrain-like data.
# Best Used For:
# - Visualizing mathematical functions of two variables
# - Understanding surfaces, slopes, and curvature
# - Scientific and engineering visualizations
# Insight:
# The surface forms a bowl shape, showing that z increases as x
# and y move further away from zero in either direction, with the
# minimum value occurring at the origin.
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
z=xx**2+yy**2

fig=plt.figure(figsize=(10,7))
ax=plt.subplot(projection='3d')
a=ax.plot_surface(xx,yy,z,cmap='winter')
fig.colorbar(a)
```

**Key Takeaways**
- Matplotlib is the low-level foundation most other Python plotting libraries build on.
- The pyplot interface (plt.*) is quick and convenient for single, simple plots.
- The object-oriented interface (fig, ax) gives full control and scales to complex, multi-panel figures.
- Choose the plot based on the type of data and the analysis goal.

# 📌 Which Matplotlib Plot Should I Use?

| **Goal** | **Recommended Plot(s)** |
|:-------------------------------|:----------------------------------------------|
| Show trends over time | `plt.plot()` |
| Compare two numerical variables | `plt.scatter()`, `plt.plot(..., 'o')` |
| Visualize a third numerical variable | `plt.scatter(s=..., c=...)` |
| Compare values across categories | `plt.bar()`, `plt.barh()` |
| Compare groups side by side | grouped `plt.bar()` with offset x positions |
| Show part-to-whole composition across categories | stacked `plt.bar(bottom=...)` |
| Visualize distribution of one variable | `plt.hist()` |
| Show simple part-to-whole relationship | `plt.pie()` |
| Label individual points | `plt.text()` |
| Mark thresholds/averages on a chart | `plt.axhline()`, `plt.axvline()` |
| Build multi-panel figures | `plt.subplots()`, `fig.add_subplot()` |
| Visualize three numerical variables | `ax.scatter3D()` |
| Visualize a path through 3D space | `ax.plot3D()` |
| Visualize a mathematical surface | `ax.plot_surface()` |
| Apply a pre-built visual theme | `plt.style.use()` |
| Export a chart to a file | `plt.savefig()` |
