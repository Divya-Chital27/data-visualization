**Importing Libraries**

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

**Importing Datasets for analysis**

```python
tips=px.data.tips()
iris=px.data.iris()
gap=px.data.gapminder()
```

**What is Plotly**
<span>

Plotly is a Python data visualization library used for creating
interactive, publication-quality graphs. Unlike static libraries,
Plotly charts support zooming, panning, hovering for details, and
animation, all with minimal code </span>

**Types of Plotly Interfaces**


*   Plotly Express (px)-<span>A high-level interface that builds an entire figure in a single function call. It works directly on tidy/long-form DataFrames and is best for quick, everyday visualizations.</span>


 - scatter()
 - line()
 - bar()
 - histogram()
 - pie()
 - sunburst()
 - treemap()
 - imshow()
 - scatter_3d()
 - scatter_matrix()



*   Graph Objects (go)-<span>A low-level interface that gives full control over every element of a figure by building it trace by trace. It is useful when a chart type or level of customization is not available through Plotly Express.</span>


- go.Scatter()
 - go.Bar()
 - go.Surface()
 - go.Contour()
 - go.Figure()
 - go.Layout()
 - make_subplots()






**Relational** **Plots**

Graph Objects-These build a plot trace by trace, giving full control over the figure.
*   Scatter Plot (go)

Plotly Express-These build the entire figure in a single function call.


*   scatter
*   line





```python
#Scatter plot using Graph Objects
# Purpose: Shows the relationship between life expectancy and GDP per capita.
# Best Used For:
# - Finding correlations
# - Detecting clusters
# - Identifying outliers
# Insight:
# Countries with higher GDP per capita generally have higher life
# expectancy. A few countries stand out as outliers with unusually
# high or low life expectancy for their income level.
temp_df=gap[gap['year']==2007]
trace1=go.Scatter(x=temp_df['lifeExp'],y=temp_df['gdpPercap'],mode='markers')
data=[trace1]
layout=go.Layout(title='lifeexp vs gdp for 2007',xaxis={'title':'life exp'},yaxis={'title':'gdp per captia'})
fig=go.Figure(data,layout)
fig.show()
```

```python
#adding multiple traces on the same figure
#a scatter trace and a line trace can be combined in a single figure
trace1=go.Scatter(x=temp_df['lifeExp'],y=temp_df['gdpPercap'],mode='markers')
trace2=go.Scatter(x=[0,1,2],y=[0,90,30000],mode='lines')
data=[trace1,trace2]
layout=go.Layout(title='lifeexp vs gdp for 2007',xaxis={'title':'life exp'},yaxis={'title':'gdp per captia'})
fig=go.Figure(data,layout)
fig.show()
```

```python
#Plotly Express-same above analysis is done using px.scatter
#color-The color parameter uses different colors to represent different categories of a variable.
#size-Represents a variable by changing the size of the markers.
#hover_name-Shows the given column as the bold title in the hover tooltip.
px.scatter(temp_df,x='lifeExp',y='gdpPercap',color='continent',size='pop',size_max=50,hover_name='country')
```

```python
#animation_frame and animation_group
# Purpose:
# Adds a play/pause slider that animates the plot across a
# variable (commonly time), letting each frame show one value
# of that variable while animation_group keeps the same marker
# tied to the same entity across frames.
# Insight:
# Playing the animation shows that most countries have moved
# up and to the right over time, indicating that both GDP per
# capita and life expectancy have generally improved worldwide,
# although some countries and continents progress faster than others.
px.scatter(gap,x='lifeExp',y='gdpPercap',color='continent',
           size='pop',size_max=50,hover_name='country',
           range_x=[30,95],
           animation_frame='year',animation_group='country')
```

```python
#line plot
# Purpose: Shows trends or changes over an ordered variable such as time.
# Best Used For:
# - Time series analysis
# - Trend analysis
# - Comparing multiple trends
#Insight:
#The graph indicates that India's population has grown steadily
# and continuously over the years.
temp_df=gap[gap['country']=='India']
px.line(temp_df,x='year',y='pop')
```

```python
#comparing multiple countries on the same line plot
# Insight:
# All three countries show a rising trend in life expectancy over
# the years, but the rate and starting point of improvement differ
# between India, China, and Pakistan.
temp_df=gap[gap.country.isin(['India','China','Pakistan'])]
temp_df=temp_df.pivot_table(index='year',columns='country',values='lifeExp')
px.line(temp_df,x=temp_df.index,y=temp_df.columns)
```

**Bar** **Plots**


*   Bar Chart
*   Grouped Bar Chart
*   Stacked Bar Chart
*   Animated Bar Chart





```python
#Bar chart
# Purpose:
# Compares values across categories using the height (or length)
# of rectangular bars.
# Best Used For:
# - Comparing quantities across categories
# - Tracking a single metric over time
# - Ranking categories
# Insight:
# India's population has increased consistently across every
# recorded year, with the steepest growth visible in the more
# recent decades.
temp_df=gap[gap['country']=='India']
px.bar(temp_df,x='year',y='pop')
```

```python
#Grouped bar chart
# Purpose:
# Places bars for different categories side by side within each
# group, making it easy to compare categories at every point.
# barmode='group'-Displays bars next to each other instead of stacked.
# log_y-Plots the y-axis on a logarithmic scale, useful when values
# differ by large magnitudes.
# text_auto-Automatically displays the value on top of each bar.
# Insight:
# China consistently has the largest population among the three
# countries, followed by India and then Pakistan, although India's
# growth rate appears to be closing the gap over time.
temp_df=gap[gap.country.isin(['India','China','Pakistan'])]
temp_df=temp_df.pivot_table(index='year',columns='country',values='pop')
px.bar(temp_df,x=temp_df.index,y=temp_df.columns,barmode='group',log_y=True,text_auto=True)
```

```python
#Stacked bar chart
# Purpose:
# Stacks bars for sub-categories on top of each other, showing
# both the total and the individual contribution of each sub-category.
# Best Used For:
# - Showing part-to-whole relationships across categories
# - Comparing totals while still showing composition
# Insight:
# Asia contributes the largest share to the world population in
# 2007, with countries like China and India forming a major portion
# of the stacked bar for that continent.
temp_df=gap[gap.year==2007]
px.bar(temp_df,x='continent',y='pop',color='country')
```

```python
#Animated bar chart
# Purpose:
# Combines a bar chart with an animation frame to show how
# category values change over time.
# Insight:
# Asia's population grows the fastest and remains the largest
# continent by population throughout the animation, while Oceania
# stays consistently the smallest.
px.bar(gap,x='continent',y='pop',color='continent',
       animation_frame='year',animation_group='country',
       range_y=[0,4000000000])
```

**Distribution Plots**

- Histogram
- Histogram with color/hue



```python
#Histogram
# Purpose:
# Displays the distribution of a single numerical variable by
# grouping data into intervals (bins) and showing the frequency
# of observations in each interval.
# Best Used For:
# - Understanding the distribution of data
# - Identifying the most common value ranges
# - Detecting skewness (left or right)
# - Checking whether data is approximately normally distributed
# text_auto-Displays the count value on top of each bar.
# Insight:
# Life expectancy across countries in 2007 is concentrated in the
# 65-80 year range, with a smaller number of countries showing
# much lower life expectancy, indicating a left-skewed distribution.
temp_df=gap[gap.year==2007]
px.histogram(temp_df,x='lifeExp',text_auto=True)
```

```python
#color
# Purpose:
# Splits the histogram into separate colored groups based on a
# categorical variable, making it easy to compare distributions.
# Insight:
# The sepal length distributions for the three iris species overlap
# but are noticeably shifted, with Setosa generally having the
# smallest sepal lengths and Virginica the largest.
px.histogram(iris,x='sepal_length',color='species')
```

**Part-of-Whole Plots**


*   Pie Chart
*   Sunburst Chart
*   Treemap





```python
#Pie chart
# Purpose:
# Shows the proportion of each category relative to the whole
# using slices of a circle.
# Best Used For:
# - Showing simple part-to-whole relationships
# - Comparing the share of a small number of categories
# Insight:
# Among European countries in 2007, a handful of countries such
# as Germany, France, and the UK make up a large share of the
# total European population, while many smaller countries each
# contribute only a thin slice.
temp_df=gap[(gap.year==2007) & (gap.continent=='Europe')]
px.pie(temp_df,values='pop',names='country')
```

```python
#pie chart on aggregated data
# Insight:
# Asia holds the largest share of world population in 1952,
# followed by Europe, while continents like Oceania contribute
# only a very small slice of the total.
temp_df=gap[gap.year==1952].groupby('continent')['pop'].sum().reset_index()
px.pie(temp_df,values='pop',names='continent')
```

```python
#Sunburst chart
# Purpose:
# Displays hierarchical data as a series of nested rings, where
# each ring represents a level in the hierarchy (e.g. continent
# then country).
# Best Used For:
# - Visualizing part-to-whole relationships across multiple levels
# - Exploring hierarchical/nested category structures
# - Comparing proportions within and across groups
# path-Defines the hierarchy levels, from outermost to innermost.
# Insight:
# The inner ring shows each continent's share of world population
# in 2007, while the outer ring breaks each continent down into
# its individual countries, making it easy to see which countries
# dominate within a continent, such as China and India within Asia.
temp_df=gap[gap['year']==2007]
px.sunburst(temp_df,path=['continent','country'],values='pop')
```

```python
#Treemap
# Purpose:
# Displays hierarchical data as nested rectangles, where the size
# of each rectangle represents its value and color can represent
# another variable.
# Best Used For:
# - Visualizing hierarchical part-to-whole relationships
# - Comparing sizes across many categories at once
# - Adding a secondary metric using color
# Insight:
# Larger rectangles correspond to countries with bigger populations,
# such as China and India, while the color scale (life expectancy)
# highlights that some densely populated countries have notably
# lower life expectancy than smaller, wealthier nations.
px.treemap(temp_df,path=[px.Constant('World'),'continent','country'],values='pop',color='lifeExp')
```

**Matrix Plots**


- Heatmap (imshow)
- Scatter Matrix




```python
#Heatmap using imshow
# Purpose:
# Visualizes values in a matrix using color intensity, making it
# easy to identify patterns, trends, and variations across two
# dimensions.
# Best Used For:
# - Comparing values across two categorical variables
# - Identifying patterns and trends
# - Detecting high and low values
# - Visualizing pivot tables and correlation matrices
# Insight:
# The heatmap compares total bill amounts across days of the week
# for male and female customers, making it easy to spot which
# day-gender combination generates the highest total spending.
temp_df=tips.pivot_table(index='day',columns='sex',values='total_bill',aggfunc='sum')
px.imshow(temp_df)
```

```python
#Scatter matrix
# Purpose:
# Displays pairwise scatter plots between multiple numerical
# variables in a single grid, similar to a pair plot.
# Best Used For:
# - Exploring relationships between multiple variables at once
# - Detecting correlations
# - Identifying clusters and patterns
# - Comparing different classes
# Insight:
# The scatter matrix reveals a strong relationship between petal
# length and petal width for the Iris dataset. Setosa forms a
# clearly distinct cluster across most feature pairs, while
# Versicolor and Virginica overlap somewhat but remain largely
# separable.
px.scatter_matrix(iris,dimensions=['sepal_length','sepal_width','petal_length','petal_width'],color='species')
```

**3D Plots**


*   3D Scatter Plot
*   3D Surface Plot (go)
*   Contour Plot (go)





```python
#3D scatter plot
# Purpose:
# Extends a regular scatter plot with a third numerical axis,
# allowing three variables to be visualized simultaneously.
# Best Used For:
# - Exploring relationships between three numerical variables
# - Detecting clusters in three dimensions
# - Identifying outliers across multiple variables at once
# Insight:
# Countries with higher GDP per capita and higher life expectancy
# tend to have smaller populations, while a few very populous
# countries stand out along the population axis regardless of
# their income or life expectancy level.
temp_df=gap[gap.year==2007]
px.scatter_3d(temp_df,x='lifeExp',y='pop',z='gdpPercap',log_y=True,color='continent')
```

```python
#3D surface plot -cannot be created using Plotly Express, built using Graph Objects
# Purpose:
# Displays a continuous surface over a two-dimensional grid,
# useful for visualizing mathematical functions or terrain-like data.
# Best Used For:
# - Visualizing mathematical functions of two variables
# - Understanding surfaces, slopes, and curvature
# - Scientific and engineering visualizations
# Insight:
# The surface forms a bowl shape, showing that the value of z
# increases as x and y move further away from zero in either
# direction, with the minimum value occurring at the origin.
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
z=xx**2+yy**2

trace1=go.Surface(x=xx,y=yy,z=z)
data=[trace1]
layout=go.Layout(title='3d surface plot')
fig=go.Figure(data,layout)
fig.show()
```

```python
#Contour plot
# Purpose:
# Represents a 3D surface in two dimensions using contour lines,
# where each line connects points of equal value.
# Best Used For:
# - Visualizing the same data as a surface plot in 2D form
# - Identifying regions of high and low value
# - Reading exact value bands more precisely than a surface plot
# Insight:
# The contour lines form concentric rings centered at the origin,
# confirming the same bowl-shaped pattern seen in the surface plot,
# with values increasing steadily as distance from the center grows.
trace1=go.Contour(x=x,y=y,z=z)
data=[trace1]
layout=go.Layout(title='contour plot')
fig=go.Figure(data,layout)
fig.show()
```

**Facet Plots**



```python
#facet plot
# Purpose:
# Creates multiple subplots by dividing the dataset based on one
# or more categorical variables, allowing easy comparison between
# groups.
# Best Used For:
# - Comparing the same visualization across categories
# - Identifying patterns within different groups
# - Exploring relationships in subsets of the data
# - Multi-panel exploratory data analysis (EDA)
# facet_col-Creates a new column of subplots for each category.
# facet_row-Creates a new row of subplots for each category.
px.scatter(tips,x='total_bill',y='tip',facet_col='sex',facet_row='smoker')
```

```python
#facet_row on a histogram
px.histogram(tips,x='total_bill',facet_row='sex')
```

```python
#facet_col_wrap
# Purpose:
# Limits the number of facet columns displayed in a single row.
# here only 4 plots are shown per row.
px.scatter(gap,x='lifeExp',y='gdpPercap',facet_col='year',facet_col_wrap=4)
```

**Subplots**



```python
#Subplots
# Purpose:
# Combines multiple independent charts (even of different types)
# into a single figure using Graph Objects, giving full control
# over the layout via rows and columns.
# Best Used For:
# - Displaying different chart types side by side
# - Building custom, multi-panel dashboards
# - Comparing unrelated metrics in one figure
# Insight:
# Placing the scatter and histogram traces side by side allows
# both charts to be viewed and compared together within a single
# figure instead of as separate plots.
fig=make_subplots(rows=1,cols=2)
fig.add_trace(
    go.Scatter(x=[1,9,5],y=[2,10,1]),
    row=1,
    col=1
)
fig.add_trace(
    go.Histogram(x=[1,9,5,22,109,134,56,789,456]),
    row=1,
    col=2
)
fig.update_layout(title='subplots demo')
fig.show()
```

**Key Takeaways**
- Plotly creates interactive, zoomable, and hoverable visualizations.
- Plotly Express (px) builds a complete figure in one function call.
- Graph Objects (go) builds a figure trace by trace for full customization.
- Choose the plot based on the type of data, number of variables, and analysis goal.

# 📌 Which Plotly Plot Should I Use?

| **Goal** | **Recommended Plot(s)** |
|:-------------------------------|:----------------------------------------------|
| Show trends over time | `px.line()` |
| Compare two numerical variables | `px.scatter()`, `go.Scatter()` |
| Animate a relationship over time | `px.scatter(animation_frame=...)` |
| Compare category totals | `px.bar()` |
| Compare category totals side by side | `px.bar(barmode='group')` |
| Show part-to-whole composition across categories | `px.bar(color=...)` |
| Animate category totals over time | `px.bar(animation_frame=...)` |
| Visualize distribution of one variable | `px.histogram()` |
| Compare distributions across categories | `px.histogram(color=...)` |
| Show simple part-to-whole relationship | `px.pie()` |
| Show hierarchical part-to-whole relationship | `px.sunburst()`, `px.treemap()` |
| Visualize a matrix / pivot table | `px.imshow()` |
| Compare relationships between many variables | `px.scatter_matrix()` |
| Visualize three numerical variables | `px.scatter_3d()` |
| Visualize a mathematical surface | `go.Surface()` |
| Visualize a surface in 2D | `go.Contour()` |
| Compare subsets of data | `facet_col`, `facet_row`, `facet_col_wrap` |
| Combine multiple chart types in one figure | `make_subplots()` |
